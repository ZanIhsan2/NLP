###**Gimana sih proses Tokenizer dari IndoBert dan Bert**

```python
# Model BERT

from transformers import AutoTokenizer

tokenizer_IndoBert = AutoTokenizer.from_pretrained("indobenchmark/indobert-base-p1")

tokenizer_Bert = AutoTokenizer.from_pretrained('bert-base-uncased')
```

```python
test_sentence = "Fasilitas di pantai ini sangat mengecewakan dan infrastrukturnya buruk."


tokens = tokenizer_IndoBert.tokenize(test_sentence)
print(tokens)
```

```python
test_sentence = test_sentence.lower()

tokens = tokenizer_IndoBert.tokenize(test_sentence)
print(tokens)
```

```python
# BERT Tokenizer
tokens_bert = tokenizer_Bert.tokenize(test_sentence)
print(tokens_bert)
```

##**Note Tugas**

coba bandingkan model encoder antara multilingual-bert dan Indobert. Buat ringkasan dalam bentuk tulisa tangan.

```python
test_sentence = test_sentence.lower()

# Tokenisasi IndoBERT
tokens_indobert = tokenizer_IndoBert.tokenize(test_sentence)
print("Tokens IndoBERT :", tokens_indobert)
print("Jumlah Token    :", len(tokens_indobert))
print("-" * 50)

# Tokenisasi mBERT
tokens_bert = tokenizer_Bert.tokenize(test_sentence)
print("Tokens mBERT    :", tokens_bert)
print("Jumlah Token    :", len(tokens_bert))
```

```python
import pandas as pd

# Membuat dataframe perbandingan jumlah token
data_perbandingan = {
    'Model': ['IndoBERT', 'Multilingual-BERT (mBERT)'],
    'Jumlah Token yang Dihasilkan': [len(tokens_indobert), len(tokens_bert)],
    'Efisiensi': ['Lebih Sedikit (Efisien)', 'Lebih Banyak (Kurang Efisien)']
}

df_compare = pd.DataFrame(data_perbandingan)
print("\nTabel Kesimpulan Efisiensi Tokenizer:")
print(df_compare)
```

```python
test_text_mix = "Hidden gem tempat wisata ini worth it banget buat healing akhir pekan."

tokens_mixed_indobert = tokenizer_IndoBert.tokenize(test_text_mix)
print("Tokens IndoBERT :", tokens_mixed_indobert)
print("Jumlah Token    :", len(tokens_mixed_indobert))
print("-" * 50)

# 3. Proses tokenisasi dengan mBERT
tokens_mixed_bert = tokenizer_Bert.tokenize(test_text_mix)
print("Tokens mBERT    :", tokens_mixed_bert)
print("Jumlah Token    :", len(tokens_mixed_bert))
```
