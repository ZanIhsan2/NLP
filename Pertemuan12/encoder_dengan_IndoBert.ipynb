##**Decoder Transformer**

karena bagian atas sudah eksplorasi dengan `encoder` based, kita bakal coba liat gimana kalo decodernya. Jadi model bahasa seperti GPT, LLAMA, itu adalah bagian transformer `Decodernya`. Tugasnya bisanya adalah memprediksi teks selanjutnya dari sebuah kalimat. Sama kayak pas kalian prompting di AI seperti ChatGPT, Gemini dan lain-lain. (Autoregressive)

```python
# model bahasa encoder GPT-2 Yang difine tuning dengan bahasa indonesia

from transformers import GPT2Tokenizer, GPT2Model

model_name = 'cahya/gpt2-small-indonesian-522M'
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
```

```python
# coba kita liat hasil tokenizernya

test_text = 'Destinasi wisata di Bali yang paling indah adalah [masked]'
encode_text = tokenizer.tokenize(test_text)
```

```python
encode_text
```

```python
model = GPT2Model.from_pretrained(model_name)

encode_input = tokenizer(test_text, return_tensors='pt')
output = model(**encode_input)
```

```python
from transformers import pipeline, set_seed
generator = pipeline('text-generation', model='cahya/gpt2-small-indonesian-522M')
set_seed(42)
```

```python
generator(test_text, max_length=3, num_return_sequences=1, num_beams=10)
```

###**Penasaran lagi nih, gimana model encoder transformer meprediksi kata selanjutnya secara probabilitynya**

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

```

```python
with torch.no_grad():
  outputs = model(**encode_input)
```

```python
skor_mentah = outputs.logits[0, -1, :]

probability = torch.nn.functional.softmax(skor_mentah, dim=-1)

top_k = torch.topk(probability, 5)
```

```python
print(test_text)
```

```python
for i in range(5):
  token_id = top_k.indices[i].item()
  prob_persen = top_k.values[i].item() * 100

  #decode kembali menjadi teks
  prediksi = tokenizer.decode([token_id])

  print(f"rangk {i} kata prediksi: {prediksi} - probabilitas: {prob_persen}")
```
