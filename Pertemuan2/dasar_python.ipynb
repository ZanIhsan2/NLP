```python
name = "Razan"
age = 19
height = 168
weight = 80
is_student = True
```

### Perulangan/Looping

```python
for i in friends:
    print(i)
```

```python
# perulangan dengan while

index = 0
while index < len(friends):
    print(friends[index])
    index += 1
```

```python
# Function(Fungsi)
def introduction(name, age, height):
    print(name)
    print(age)
    print(height)
```

```python
introduction("Razan", 19, 168)
```
